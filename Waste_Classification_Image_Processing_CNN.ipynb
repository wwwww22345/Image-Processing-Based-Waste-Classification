## 6. Image Processing Functions

```python
def original_preprocess(image):
    """
    Baseline preprocessing:
    Only applies MobileNetV2 preprocessing.
    which normalizes pixel values into the format expected by the pretrained network.
    """
    image = image.astype(np.float32)
    image = preprocess_input(image)
    return image


```

```python
def median_filter_preprocess(image):
    """
    This function applies a Median Filter with a kernel size of 3×3 
    to reduce impulse noise.
    After filtering, the image is normalized using MobileNetV2 preprocessing.
    """
    image = image.astype(np.uint8)
    image = cv2.resize(image, (224, 224))
    image = cv2.medianBlur(image, 3)
    image = preprocess_input(image.astype(np.float32))
    return image


```

```python
def gaussian_filter_preprocess(image):
    """
    Applies Gaussian Filter using a 3×3 kernel to smooth minor variations.
    After filtering, the image is normalized using MobileNetV2 preprocessing.
    """
    image = image.astype(np.uint8)
    image = cv2.resize(image, (224, 224))
    image = cv2.GaussianBlur(image, (3, 3), 0)
    image = preprocess_input(image.astype(np.float32))
    return image


```

```python
def unsharp_preprocess(image):
    """
    Applies Unsharp Masking to enhance edges.
    A blurred version of the image is subtracted from the original image
    and then combined using weighted addition.
    MobileNetV2 preprocessing is applied after sharpening.
    """
    image = image.astype(np.uint8)
    image = cv2.resize(image, (224, 224))

    blurred = cv2.GaussianBlur(image, (5, 5), 1)
    sharpened = cv2.addWeighted(image, 1.5, blurred, -0.5, 0)

    image = preprocess_input(sharpened.astype(np.float32))
    return image


```

```python
def highpass_preprocess(image):
    """
    Applies High-pass Filtering to strongly enhance details.
    The method amplifies high-frequency components by combining the original 
    image with a negatively weighted blurred version, producing stronger 
    sharpening effects than standard unsharp masking
    After filtering, the image is normalized using MobileNetV2 preprocessing.
    """
    image = image.astype(np.uint8)
    image = cv2.resize(image, (224, 224))

    blurred = cv2.GaussianBlur(image, (5, 5), 1)
    highpass = cv2.addWeighted(image, 2.0, blurred, -1.0, 0)

    image = preprocess_input(highpass.astype(np.float32))
    return image


```

```python
def histogram_equalization_preprocess(image):
    """
    Applies Histogram Equalization on the brightness channel by redistributing 
    pixel intensity values. The operation is performed on the brightness 
    channel in the YCrCb color space, 
    YCbCr: is a family of color spaces used as a part of the color image pipeline in digital video and photography
    It helps enhance visibility in darker and 
    brighter regions while preserving color information
    """
    image = image.astype(np.uint8)
    image = cv2.resize(image, (224, 224))

    ycrcb = cv2.cvtColor(image, cv2.COLOR_RGB2YCrCb)
    ycrcb[:, :, 0] = cv2.equalizeHist(ycrcb[:, :, 0])
    enhanced = cv2.cvtColor(ycrcb, cv2.COLOR_YCrCb2RGB)

    image = preprocess_input(enhanced.astype(np.float32))
    return image


```

```python
def gaussian_canny_preprocess(image):
    """
    Applies Gaussian Filter followed by Canny edge enhancement to highlight object boundaries while reducing image noise.
    first applies a Gaussian blur to reduce noise, then uses the Canny edge detector to find edges.
    The resulting edge map is converted to RGB and blended with the original image to enhance details.
    After enhancement, the image is normalized using MobileNetV2 preprocessing.
    """
    image = image.astype(np.uint8)
    image = cv2.resize(image, (224, 224))

    blurred = cv2.GaussianBlur(image, (3, 3), 0)
    gray = cv2.cvtColor(blurred, cv2.COLOR_RGB2GRAY)
    edges = cv2.Canny(gray, 100, 200)
    edges_rgb = cv2.cvtColor(edges, cv2.COLOR_GRAY2RGB)

    enhanced = cv2.addWeighted(blurred, 0.85, edges_rgb, 0.15, 0)

    image = preprocess_input(enhanced.astype(np.float32))
    return image


```

```python
def morphology_preprocess(image):
    """
    Applies thresholding, opening, and closing.
    Tested as morphological preprocessing.
    to segment the main waste object from the background. 
    Opening removes small noise regions, while closing fills small gaps and smooths object boundaries.
    first converts the image to grayscale, applies Otsu's thresholding to create a binary mask, 
    and then uses morphological opening and closing to refine the mask. 
    The final segmented image is obtained by applying the mask to the original image, which is then normalized using MobileNetV2 preprocessing.
    """
    image = image.astype(np.uint8)
    image = cv2.resize(image, (224, 224))

    gray = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)

    _, binary = cv2.threshold(
        gray,
        0,
        255,
        cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU
    )

    kernel = np.ones((5, 5), np.uint8)

    opened = cv2.morphologyEx(binary, cv2.MORPH_OPEN, kernel)
    closed = cv2.morphologyEx(opened, cv2.MORPH_CLOSE, kernel)

    segmented = cv2.bitwise_and(image, image, mask=closed)

    image = preprocess_input(segmented.astype(np.float32))
    return image
```

```python
def get_six_sample_images():
    sample_images = []

    for class_name in classes:
        class_path = os.path.join(dataset_dir, class_name)
        image_name = os.listdir(class_path)[0]
        image_path = os.path.join(class_path, image_name)

        image = cv2.imread(image_path)
        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        image = cv2.resize(image, (224, 224))

        sample_images.append((class_name, image))

    return sample_images


sample_images = get_six_sample_images()
```

### Apply Methods for Display

```python
def apply_display_method(image, method_name):
    '''
    This function applies each preprocessing method to sample images for visual comparison and histogram analysis. 
    It is used for display only and is not part of the model training process.
    '''
    
    if method_name == "Original":
        return image

    elif method_name == "Gaussian Filter":
        return cv2.GaussianBlur(image, (3, 3), 0)

    elif method_name == "Median Filter":
        return cv2.medianBlur(image, 3)

    elif method_name == "Unsharp Masking":
        blurred = cv2.GaussianBlur(image, (5, 5), 1)
        return cv2.addWeighted(image, 1.5, blurred, -0.5, 0)

    elif method_name == "High-Pass Filtering":
        blurred = cv2.GaussianBlur(image, (5, 5), 1)
        return cv2.addWeighted(image, 2.0, blurred, -1.0, 0)

    elif method_name == "Histogram Equalization":
        ycrcb = cv2.cvtColor(image, cv2.COLOR_RGB2YCrCb)
        ycrcb[:, :, 0] = cv2.equalizeHist(ycrcb[:, :, 0])
        return cv2.cvtColor(ycrcb, cv2.COLOR_YCrCb2RGB)

    elif method_name == "Gaussian + Canny":
        blurred = cv2.GaussianBlur(image, (3, 3), 0)
        gray = cv2.cvtColor(blurred, cv2.COLOR_RGB2GRAY)
        edges = cv2.Canny(gray, 100, 200)
        edges_rgb = cv2.cvtColor(edges, cv2.COLOR_GRAY2RGB)
        return cv2.addWeighted(blurred, 0.85, edges_rgb, 0.15, 0)

    return image
```

## 8. Create Data Generators

```python
''' 
creates the data generators used for training and validation. 
The ImageDataGenerator automatically loads images from the dataset directory, 
applies the selected preprocessing method, and splits the dataset into 80% 
training data and 20% validation data. 

Images are resized to 224 × 224 pixels, 
grouped into batches of 32 images, and converted into categorical labels suitable 
for multi-class classification. 

The training generator shuffles the data to improve learning, 
while the validation generator keeps the data order fixed for consistent evaluation.
'''
img_size = (224, 224)
batch_size = 32
validation_split = 0.2
''' 
the train_generator loads images from the dataset directory, applies the 
specified preprocessing function, resizes the images to 224x224 pixels, groups 
them into batches of 32, and assigns categorical labels based on the directory 
structure. It uses 80% of the data for training and shuffles the data to improve learning.

the val_generator loads the same images but uses the remaining 20% of the data 
for validation.
'''
def create_generators(preprocess_function):
    datagen = ImageDataGenerator(
        preprocessing_function=preprocess_function,
        validation_split=validation_split
    )
    
    train_generator = datagen.flow_from_directory(
    dataset_dir,
    target_size=img_size,
    batch_size=batch_size,
    class_mode="categorical",
    subset="training",
    shuffle=True,
    seed=42
)

    val_generator = datagen.flow_from_directory(
    dataset_dir,
    target_size=img_size,
    batch_size=batch_size,
    class_mode="categorical",
    subset="validation",
    shuffle=False,
    seed=42
    
)
    return train_generator, val_generator
```

## 9. Build MobileNetV2 Model

```python
''' 
builds a convolutional neural network model using MobileNetV2 as the base architecture. 
The base model is loaded with pretrained weights from ImageNet and is set to non-trainable to leverage its feature extraction capabilities without modifying its weights. 
On top of the base model, a Global Average Pooling layer is added to reduce the spatial dimensions, followed by a Dropout layer with a rate of 0.3 to prevent overfitting. 
A Dense layer with 128 units and ReLU activation is added for learning complex patterns, and a final Dense layer with 6 units and softmax activation is used for multi-class classification corresponding to the six waste classes. 
The model is compiled with the Adam optimizer, categorical cross-entropy loss function, and accuracy as the evaluation metric, making it ready for training on the waste classification task.
'''

def build_model():
    base_model = MobileNetV2(
        weights="imagenet",
        include_top=False,
        input_shape=(224, 224, 3)
    )

    base_model.trainable = False

    model = Sequential([
        base_model,
        GlobalAveragePooling2D(),
        Dropout(0.3),
        Dense(128, activation="relu"),
        Dense(6, activation="softmax")
    ])

    model.compile(
        optimizer="adam",
        loss="categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model
```

## 10. Train and Evaluate Function

```python
'''  
trains and evaluates a model using the specified preprocessing method.
'''
def train_and_evaluate(method_name, preprocess_function, epochs=10):
    print(f"\n========== Training: {method_name} ==========")

    train_gen, val_gen = create_generators(preprocess_function)

    model = build_model()

    history = model.fit(
        train_gen,
        validation_data=val_gen,
        epochs=epochs
    )

    final_accuracy = history.history["val_accuracy"][-1]

    print(f"\n{method_name} Final Validation Accuracy: {final_accuracy:.4f}")

    return {
        "method": method_name,
        "model": model,
        "history": history,
        "val_generator": val_gen,
        "accuracy": final_accuracy
    }
```
